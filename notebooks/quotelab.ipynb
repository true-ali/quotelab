# QuoteLab: reproduce the report

This notebook rebuilds the report of one dataset from the files `quotelab backtest` wrote under `runs/<dataset>/`, then shows the main tables and every figure. It runs no backtest itself.

Before running it:

- synthetic data: `make backtest` (about a minute);
- Binance data: `make data` and `make backtest-real` (hours), then set `DATASET = "real"` below.

Running it rewrites `reports/<dataset>/` and that dataset's block in `README.md`, exactly like `make report`. Everything here is a simulation and none of it is investment advice.

```python
DATASET = "synthetic"
```

```python
from IPython.display import Image, Markdown, display

from quotelab.config import load_config, resolve_path
from quotelab.report.results import report_dir, write_report
from quotelab.report.sections import FIGURES
from quotelab.report.tables import load_outputs, report_tables

cfg = load_config()
results = write_report(cfg, DATASET)
tables = report_tables(load_outputs(resolve_path(cfg["experiments"]["runs_dir"]) / DATASET), cfg)
print(results)
```

## E1 realism ladder, E2 strategies and the E7 break-even fee

Mean daily PnL in USD with 95% block-bootstrap intervals over test days.

```python
tables.realism.select(
    "variant", "symbol", "n_days", "pnl_mean", "pnl_lo", "pnl_hi", "change_vs_previous"
)
```

```python
tables.strategies.select(
    "variant",
    "symbol",
    "pnl_mean",
    "attr_spread",
    "attr_inventory",
    "attr_fees",
    "quote_fills",
    "clamp_share",
)
```

```python
tables.break_even
```

## Figures

```python
for anchor, relative in FIGURES.items():
    display(Markdown(f"### {anchor}"))
    display(Image(filename=str(report_dir(cfg, DATASET) / relative)))
```

The full write-up, with every table and the text generated from it, is `reports/<dataset>/RESULTS.md`; `reports/<dataset>/dashboard.html` shows the same report as a page.
